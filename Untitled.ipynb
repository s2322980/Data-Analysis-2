```julia
using XLSX
using DataFrames
using CairoMakie

# ========== SETTINGS ==========
filename = "Raman_data.xlsx"
sheetname = "wt"   # Change to "rtcA", "rtcB", or "rtcR" to plot others
growth_rates = [0.3, 0.4, 0.5, 0.7, 0.9, 0.6, 0.8, 1.0]  # Example λ for each cell (normalize 0–1)

# ========== READ DATA ==========
sheet = XLSX.readtable(filename, sheetname)

# Convert sheet to DataFrame for convenience
df = DataFrame(sheet...)

# Extract biomolecule names
biomolecules = df[:, 1]

# Find indices for the biomolecules we need
idx_protein = findfirst(==("prt"), biomolecules)
idx_rna     = findfirst(==("rna"), biomolecules)
idx_ratio   = findfirst(x -> occursin("RNA/Protein", lowercase(String(x))), biomolecules)

# Extract intensities for all cells (columns from 2 onward)
cols = names(df)[2:end]
protein_vals = [df[idx_protein, c] for c in cols]
rna_vals     = [df[idx_rna, c] for c in cols]
ratio_vals   = [df[idx_ratio, c] for c in cols]

# ========== 3D SCATTER PLOT ==========
fig = Figure(resolution = (800, 600))
ax = Axis3(fig[1, 1],
    title = "RNA vs Protein vs Ratio (wt, no tetracycline)",
    xlabel = "Protein intensity",
    ylabel = "RNA intensity",
    zlabel = "RNA/Protein ratio")

# Normalize growth rates for color mapping (optional)
λnorm = (growth_rates .- minimum(growth_rates)) ./ (maximum(growth_rates) - minimum(growth_rates))

# Scatter with color indicating growth rate
scatter!(ax, protein_vals, rna_vals, ratio_vals, markersize = 12,
         color = λnorm, colormap = :viridis)

Colorbar(fig[1, 2], limits = (0, 1), colormap = :viridis,
         label = "Normalized growth rate (λ)")

fig
save("rna_protein_ratio_3D_wt.png", fig)
```
